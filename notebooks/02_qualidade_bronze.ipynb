```python
import builtins
from pyspark.sql import functions as F

#Verificar valores ausentes representados por "-", vazio ou NULL

for tabela in tabelas_bronze:
    df = spark.table(f"mvp_1.bronze.{tabela}")

    #Considera somente as colunas que representam anos
    colunas_anos = [c for c in df.columns if c.isdigit()]

    expressoes = [
        F.sum(
            F.when(
                F.col(c).isNull() |
                (F.col(c) == "") |
                (F.col(c) == "-"),
                1
            ).otherwise(0)
        ).alias(c)
        for c in colunas_anos
    ]

    resultado = df.select(expressoes).collect()[0].asDict()

    #Soma
    total_ausentes = builtins.sum(resultado.values())

    print(f"{tabela}: {total_ausentes} valores ausentes")
```

Verificação de valores ausentes

Verificamos valores NULL, strings vazias e valores representados por - nas colunas anuais das nove tabelas.

Não foram encontrados valores ausentes nesses formatos.

Os valores 0 foram mantidos como valores válidos.

```python
from pyspark.sql import functions as F

# Verificar valores de cobertura fora do intervalo 0–100

for tabela in tabelas_bronze:
    df = spark.table(f"mvp_1.bronze.{tabela}")

    colunas_anos = [c for c in df.columns if c.isdigit()]

    # Converte temporariamente os valores para número
    # substituindo vírgula decimal por ponto
    valores = []

    for ano in colunas_anos:
        valores.append(
            F.regexp_replace(F.col(ano), ",", ".")
             .cast("double")
             .alias("cobertura")
        )

    # Junta todas as colunas anuais em uma única coluna
    df_valores = (
        df.select(F.explode(F.array(*valores)).alias("cobertura"))
    )

    negativos = df_valores.filter(F.col("cobertura") < 0).count()
    acima_100 = df_valores.filter(F.col("cobertura") > 100).count()

    estatisticas = (
        df_valores
        .agg(
            F.min("cobertura").alias("minimo"),
            F.max("cobertura").alias("maximo")
        )
        .collect()[0]
    )

    print(
        f"{tabela}: "
        f"negativos={negativos} | "
        f"acima_100={acima_100} | "
        f"mínimo={estatisticas['minimo']} | "
        f"máximo={estatisticas['maximo']}"
    )
```

Não foram identificados registros geográficos duplicados nas nove tabelas da camada Bronze.

```python
from pyspark.sql import functions as F

for tabela in ["tvd1_municipio", "tvd2_municipio", "scrvz_municipio"]:
    
    df = spark.table(f"mvp_1.bronze.{tabela}")
    coluna_geo = df.columns[0]
    colunas_anos = [c for c in df.columns if c.isdigit()]

    # Transforma temporariamente tipo de dado
    pares = []

    for ano in colunas_anos:
        pares.extend([
            F.lit(ano),
            F.regexp_replace(F.col(ano), ",", ".").cast("double")
        ])

    df_long = (
        df.select(
            F.col(coluna_geo).alias("local"),
            F.create_map(*pares).alias("valores")
        )
        .select(
            "local",
            F.explode("valores").alias("ano", "cobertura")
        )
    )

    print(f"\n{tabela}")

    display(
        df_long
        .orderBy(F.col("cobertura").desc())
        .limit(10)
    )
```

Não foram identificados valores negativos, o que era esperado. Porém, foram encontrados valores superiores a 100% em determinados município e anos, não sendo restritos a linha TOTAL que já apresentava esse problema. Considerando que a cobertura vacinal é um cálculo da razão do número de doses aplicadas pela população do município, esses valores superiores a 100% não foram considerados erros.
