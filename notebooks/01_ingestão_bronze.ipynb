```python
# Problema estava no que o nome da coluna total tinha um espaço
# Remove espaços no início e no fim dos nomes das colunas
####Durante a ingestão, foi identificado um espaço em branco no início do nome da coluna Total, incompatível com as regras de nomenclatura do Delta Lake. Foi aplicado strip() aos nomes das colunas antes da persistência, sem alteração dos valores originais.

df_teste = df_teste.toDF(*[coluna.strip() for coluna in df_teste.columns])

# Conferência
for coluna in df_teste.columns:
    print(repr(coluna))
```

```python
#Função para leitura e gravação dos arquivos
def carregar_bronze(nome_arquivo, nome_tabela):
    caminho = f"/Volumes/mvp_1/bronze/dados_brutos/{nome_arquivo}"

    #Leitura do arquivo original
    df = (
        spark.read
        .option("header", "true")
        .option("inferSchema", "false")
        .option("sep", ";")
        .option("encoding", "ISO-8859-1" )
        .csv(caminho)
    )

    #Padronização técnica dos nomes das colunas
    #Remove espaços das extremidades e substitui espaços internos por "_"
    novos_nomes = [
        coluna.strip().replace(" ", "_")
        for coluna in df.columns
    ]

    df = df.toDF(*novos_nomes)
    #Gravação como tabela Delta
    (
        df.write
        .format("delta")
        .mode("overwrite")
        .saveAsTable(f"mvp_1.bronze.{nome_tabela}")
    )

    print(f"Tabela {nome_tabela} criada com sucesso.")
```

```python
# Arquivos restantes que serão carregados

arquivos_bronze = {
    "TVD1_ANO_UF.csv": "tvd1_uf",
    "TVD1_ANO_REGIAO.csv": "tvd1_regiao",

    "TVD2_ANO_MUNICIPIO.csv": "tvd2_municipio",
    "TVD2_ANO_UF.csv": "tvd2_uf",
    "TVD2_ANO_REGIAO.csv": "tvd2_regiao",

    "Tetraviral_SCRVZ_ANO_MUNICIPIO.csv": "scrvz_municipio",
    "Tetraviral_SCRVZ_ANO_UF.csv": "scrvz_uf",
    "Tetraviral_SCRVZ_ANO_REGIAO.csv": "scrvz_regiao"
}

for arquivo, tabela in arquivos_bronze.items():
    carregar_bronze(arquivo, tabela)
```
